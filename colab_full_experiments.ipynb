```python
#@title Upload checkpoint (run after reconnecting)
from google.colab import files
import zipfile
import os

uploaded = files.upload()
if 'checkpoints.zip' in uploaded:
    with zipfile.ZipFile('checkpoints.zip', 'r') as zip_ref:
        zip_ref.extractall('src/env/')
    print('Checkpoint restored!')
else:
    print('No checkpoint uploaded, will train from scratch')
```
