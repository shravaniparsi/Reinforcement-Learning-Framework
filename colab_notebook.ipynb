```python
# Display results summary
import json

with open('src/env/results/full_experiment_results.json', 'r') as f:
    results = json.load(f)

print('=== Results Summary ===')
for strategy, data in results.items():
    if isinstance(data, list) and data:
        mean_reward = sum(d.get('mean_reward', 0) for d in data) / len(data)
        print(f'{strategy}: {mean_reward:.4f}')
```
